```python
# @title 2. Subir archivo PDF
from google.colab import files
import os

uploaded = files.upload()
pdf_filename = list(uploaded.keys())[0]

print(f"✅ Archivo '{pdf_filename}' subido con éxito.")
```

```python
# @title 3. Convertir a Markdown (.md)
from docling.document_converter import DocumentConverter

print("⏳ Convirtiendo archivo... (esto puede tardar unos segundos)")

converter = DocumentConverter()
result = converter.convert(pdf_filename)
markdown_output = result.document.export_to_markdown()

# Guardar el resultado en un archivo
output_filename = os.path.splitext(pdf_filename)[0] + ".md"
with open(output_filename, "w", encoding="utf-8") as f:
    f.write(markdown_output)

print(f"✅ Conversión finalizada: {output_filename}")
```
