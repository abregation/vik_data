```python
import fitz  # PyMuPDF
import os
from google.colab import files

def extract_annotations(pdf_path):
    """
    Extracts highlights and comments from a PDF file.
    Returns a string formatted in Markdown.
    """
    doc = fitz.open(pdf_path)
    markdown_output = f"# Annotations for {os.path.basename(pdf_path)}\n\n"

    found_any = False

    for page_num, page in enumerate(doc):
        annots = page.annots()
        if not annots:
            continue

        page_header_added = False

        for annot in annots:
            # annot.type: 8 is Highlight, 9 is Underline, 10 is Squiggly, 11 is Strikeout
            # annot.type: 0 is Text (Sticky Note)

            subtype = annot.type[0]
            info = annot.info
            content = info.get("content", "").strip() # This is the user's comment

            # Extract highlighted text
            # We use the annotation's coordinates to "clip" the text from the page
            highlighted_text = ""
            if subtype == 8: # Highlight
                # Get the text contained within the highlight's rectangles
                highlighted_text = page.get_text("text", clip=annot.rect).replace("\n", " ").strip()

            if highlighted_text or content:
                if not page_header_added:
                    markdown_output += f"- Page {page_num + 1}\n\n"
                    page_header_added = True

                found_any = True

                if highlighted_text:
                    markdown_output += f"- *Highlight:* {highlighted_text}\n\n"

                if content:
                    markdown_output += f"- **Comment:** {content}\n\n"

    if not found_any:
        return "No highlights or comments found in this PDF."

    return markdown_output

def run_app():
    print("1. Please upload your PDF file.")
    uploaded = files.upload()

    for filename in uploaded.keys():
        print(f"Processing '{filename}'...")

        # 2 & 3. Identify highlights and comments
        md_text = extract_annotations(filename)

        # 4. Display the text in Markdown (Preview)
        print("\n--- GENERATED MARKDOWN ---\n")
        print(md_text)

        # 5. Allow to export
        export_name = filename.rsplit('.', 1)[0] + "_annotations.md"
        with open(export_name, "w") as f:
            f.write(md_text)

        print(f"\n--- Done! ---")
        files.download(export_name)

# Run the process
run_app()
```
