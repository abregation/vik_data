```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- 1. Selector for the initial number ---
num_input = widgets.FloatText(
    value=100.0,  # Default value
    description='Enter Number:',
    disabled=False,
    style={'description_width': 'initial'} # Ensure full description is visible
)

# --- 2. Selector for the divisor ---
divisor_input = widgets.FloatText(
    value=2.0,    # Default value, cannot be 0 initially
    description='Divide by:',
    disabled=False,
    style={'description_width': 'initial'}
)

# --- 3. Selector for the multiplier ---
multiplier_options = [
    ('1,000', 1000),
    ('10,000', 10000),
    ('100,000', 100000),
    ('1,000,000', 1000000)
]
multiplier_dropdown = widgets.Dropdown(
    options=multiplier_options,
    value=1000, # Default multiplier
    description='Multiply Result by:',
    disabled=False,
    style={'description_width': 'initial'}
)

# --- Output area to display results ---
output_area = widgets.Output()

# --- Function to perform calculation and update output ---
def calculate_and_display(change): # 'change' is a dictionary passed by the observer
    with output_area:
        clear_output(wait=True) # Clear previous output

        num_val = num_input.value
        div_val = divisor_input.value
        mult_val = multiplier_dropdown.value

        print(f"Initial Number: {num_val}")
        print(f"Dividing by: {div_val}")

        if div_val == 0:
            print("Error: Cannot divide by zero.")
            result_division = "N/A (Division by zero)"
            final_result_str = "N/A"
        else:
            try:
                result_division = num_val / div_val
                final_result = result_division * mult_val
                print(f"Result of Division ({num_val}/{div_val}): {result_division:.4f}") # Format to 4 decimal places
                final_result_str = f"{final_result:,.4f}" # Format with commas and 4 decimal places
            except Exception as e:
                print(f"An error occurred during calculation: {e}")
                result_division = "Error"
                final_result_str = "Error"

        print(f"Multiplying by: {mult_val:,}") # Format multiplier with commas
        print("---")
        print(f"Final Result: {final_result_str}")


# --- Link widgets to the calculation function ---
# Call calculate_and_display whenever the 'value' of any input widget changes
num_input.observe(calculate_and_display, names='value')
divisor_input.observe(calculate_and_display, names='value')
multiplier_dropdown.observe(calculate_and_display, names='value')

# --- Display the widgets ---
print("Interactive Calculator:")
display(num_input)
display(divisor_input)
display(multiplier_dropdown)
display(output_area)

# --- Initial calculation and display ---
calculate_and_display(None) # Call once to show initial result
```
